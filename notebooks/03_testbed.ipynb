```python
import pandas as pd
from pathlib import Path

data_dir = Path("../data")
settings_file = data_dir / "settings.csv"

print("Current folder:", Path.cwd())
print("Does settings.csv exist?", settings_file.exists())

if settings_file.exists():
    settings_df = pd.read_csv(settings_file, header=None, index_col=0).squeeze("columns")
    BASE_DATE = settings_df["base_date"]
    print("Success! Base date loaded:", BASE_DATE)
else:
    print("Error: settings.csv is missing from the data folder.")
```
